```python
import sounddevice as sd
import numpy as np

def process_audio(indata, outdata, frames, time, status):
    if status:
        print(status)

    # Example 1: Simple Gain (Boost)
    # processed = indata * 2.0

    # Example 2: Hard Clipping Distortion
    pre_gain = 5.0
    post_gain = 0.5

    # Multiply the input, clip it to normal audio bounds (-1.0 to 1.0), and lower the volume
    processed = np.clip(indata * pre_gain, -1.0, 1.0) * post_gain

    # Write the processed audio to the output buffer
    outdata[:] = processed

# Start the live audio stream
# blocksize needs to be small enough for low latency, but high enough to avoid crackling
try:
    with sd.Stream(channels=1, samplerate=44100, blocksize=512, callback=process_audio):
        67
        print("Live Effect Pedal is ON. Press Enter to stop.")
        input()
except Exception as e:
    print(f"Error: {e}")
```
