```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import soundfile
```

```python
def cut_audio(audio, sr, start_time, end_time):
    """
    Cuts a segment of audio from the given audio signal.

    Parameters:
    audio (numpy.ndarray): The input audio signal.
    sr (int): The sample rate of the audio signal.
    start_time (float): The start time of the segment to cut (in seconds).
    end_time (float): The end time of the segment to cut (in seconds).

    Returns:
    numpy.ndarray: The cut audio segment.
    """
    start_sample = int(start_time * sr)
    end_sample = int(end_time * sr)

    return audio[start_sample:end_sample]


def Cut_Audio(audio_path, save_path, start_time=130.0, end_time=140.0):
    audio, sr = librosa.load(audio_path, sr=None)

    cut_segment = cut_audio(audio, sr, start_time, end_time)
    # Save the cut segment to a new file
    soundfile.write(save_path, cut_segment, sr)

save1 = 'Data/ML Guitar_ Initial_Data_Cut.wav'
input1 = 'Data/ML Guitar_ Initial_Data.wav'
Cut_Audio(input1, save1, start_time=130.0, end_time=140.0)
save2 = 'Data/ML Guitar_ Target Data_Cut.wav'
input2 = 'Data/ML Guitar Target Data.wav'
Cut_Audio(input2, save2, start_time=130.0, end_time=140.0)
```
